##### Facebook AI Similarity Search (FAISS) is a library for efficient similarity search and clustering of desne vectors. ti contain algorithms that search in sets of vectors of any size, up to ones that possibly do not fit in RAm. it also contains supporting code for evaluation and oarameter tuning.

```python
from  langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_ollama import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

loader=TextLoader("C:\\Users\\rvriksh\\GenAI\\Data_Ingestion\\speech.txt")
document=loader.load()
text_splitter=CharacterTextSplitter(chunk_size=100,chunk_overlap=0)
docs=text_splitter.split_documents(document)
```

```python
docs
```

```python
 
embeddings=OllamaEmbeddings(model="nomic-embed-text")
db=FAISS.from_documents(docs,embeddings) # document and embeddings
db
```

```python
# start querying in vector database
query=" star wall love work?"
db.similarity_search(query)[0].page_content  # this is similarity seach 
```

```python
## reterievel method of search
##we can convert the vector store into retriever class. this allow us to easily use it in other langchain methods  which largely work with retrievers
retriever=db.as_retriever()
doc=retriever.invoke(query)
doc[0].page_content
```

```python
## similarity search with score in FAISS 
## similarity_search_with_score  which allow you to return not only the document but also the distance score of the query , lower the distance better the result 

doc_with_score=db.similarity_search_with_score(query)
doc_with_score # always return  the most relevant document first in the answer list

for d in doc_with_score:
    print(f"Score: {d[1]}, Document: {d[0].page_content}")
```

```python
# can we pass the vector instead of the query to get the result 
# we can pass the vecotor , we can pass the query to search from vector database
embed_vector=embeddings.embed_query(query)
 
db=db.similarity_search_by_vector(embed_vector)
db[0].page_content
```

```python
# saving and loading into the local machine
db.save_local("faiss_index")

# load the  local save file 
new_db = FAISS.load_local(
    "faiss_index",
    embeddings,
    allow_dangerous_deserialization=True
)
new_db.similarity_search(query)
```
