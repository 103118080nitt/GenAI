```python
# building a sample vectordb
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("C:\\Users\\rvriksh\\GenAI\\Data_Ingestion\\speech.txt")
data=loader.load()

data
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=0)
splits=text_splitter.split_documents(data)
splits
```

```python
embedding= OllamaEmbeddings(model="nomic-embed-text")
vectordb=Chroma.from_documents(documents=splits,embedding=  embedding)
 
```

```python
query="Talk cold"
result=vectordb.similarity_search(query)
result[0].page_content
```

```python
vectordb=Chroma.from_documents(documents=splits,embedding=  embedding,persist_directory="./chromaDB")
```

```python
# access the save directory 
db2=Chroma(persist_directory="./chromaDB",embedding_function=embedding)

db2.similarity_search(query)[0].page_content
## we can write to search according to similarity score.

db2.similarity_search_with_score(query)[0]
```

```python
## retriever option, it allow us to use the retriever in any method.
retriver=db2.as_retriever()
retriver.invoke(query)[0].page_content
```
