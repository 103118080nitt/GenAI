```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model="gpt-4o")
llm
```

```python
## now we have llm model, let's chatt with it
result=llm.invoke("what is GenerativeAI?")
print(result.content)
```

```python
## chatprompt template

from langchain_core.prompts import ChatPromptTemplate 

prompt=ChatPromptTemplate.from_messages([
    ("system","you are an expert AI engineer , provide me answer based on questions"),
    ("user","{input}")
])
prompt
```

```python
chain=prompt|llm
response=chain.invoke({"input":"can you tell me about langsmith?"})
print(response)
```

```python
# print(type(response))
## string output parser   -- getting message by llm and how do you want to display it. 
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()

chain=prompt|llm|output_parser
 
response=chain.invoke({"input":"can you tell me about langsmith?"})
print(response)

```
