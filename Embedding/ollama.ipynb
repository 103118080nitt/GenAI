ollama support embedding models, making it possible to build reterievel augmented generation (RAG) applications that combine text prompts with existing documents or other data



```python
from langchain_ollama import OllamaEmbeddings
#from langchain_community.embeddings import OllamaEmbeddings
```

```python
embedding = OllamaEmbeddings(model="nomic-embed-text")
embedding
```

```python
embedding.embed_documents([
    "Alpha is the first letter of the greek alphabet.",
    "Beta is the second letter of the greek alphabet.",
    "Gamma is the third letter of the greek alphabet"
    ])
 
```

```python
# embbed query

embedding.embed_query("Alpha is the first letter of the greek alphabet.")
 
```

```python
### other embedding models available in ollama
 #https://ollama.com/blog/embedding-models

embed =OllamaEmbeddings(model="mxbai-embed-large")  
text="this is a test documents"
query=embed.embed_query(text)
query
```
