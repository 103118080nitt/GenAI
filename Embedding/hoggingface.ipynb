```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
text = """
Artificial intelligence enables computers to perform tasks that normally require
human intelligence, such as understanding language, recognizing images, and
making predictions.

Machine learning is a branch of artificial intelligence that learns patterns
from data. Deep learning uses neural networks with multiple layers to process
complex information.

Embeddings represent text as numerical vectors. Similar meanings produce vectors
that are close together, which helps applications perform semantic search,
recommendations, and document classification.
"""

query=embedding.embed_query(text)
query[0]
```

```python
doc_result=embedding.embed_documents([text,"this is not a test document"])
doc_result
```
